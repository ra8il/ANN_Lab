```python
import numpy as np

x = np.array([[0,0],
             [0,1],
             [1,0],
             [1,1]])

y = np.array([0,0,0,1])

weight = [0.1, 0.1]

bias = 0.1

learning_rate = 0.1

epochs = 10
    
def activation(X):
    if(X>=0):
        return 1
    else:
        return 0
    
for epoch in range(epochs):
    print(f"Epochs {epoch + 1}")
    for i in range(len(x)):
        net_input = np.dot(x[i], weight) + bias
        print(net_input)
        pred = activation(net_input)
        print(pred)
        error = y[i] - pred
        weight = weight - learning_rate * error * x[i]
        bias = bias + learning_rate * error
        print(f"Input {x[i]}, Predicted {pred}, Actual {y[i]}")
    
# error = actual_val - observed_val
```

```python
import numpy as np
import  matplotlib.pyplot as plt

x = np.linspace(-10,10,400)

def step(x):
    if(x>=0):
        return 1
    else:
        return 0
    
y = []
for var in x:
    y.append(step(var))
    
plt.plot(x,y)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Step Activaton function: ")
plt.grid()
plt.show()
```

```python
import numpy as np
import  matplotlib.pyplot as plt

x = np.linspace(-10,10,400)

def sigmoid(x):
        return 1/(1+np.exp(-x))
    
y = []
for var in x:
    y.append(sigmoid(var))
    
plt.plot(x,y)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Sigmoid Activaton function: ")
plt.grid()
plt.show()
```

```python
import numpy as np
import  matplotlib.pyplot as plt

x = np.linspace(-10,10,400)

def tanh(x):
        return (np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x))
    
y = []
for var in x:
    y.append(tanh(var))
    
plt.plot(x,y)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Tanh Activaton function: ")
plt.grid()
plt.show()
```

```python
import numpy as np
import  matplotlib.pyplot as plt

x = np.linspace(-10,10,400)

def relu(x):
        return np.maximum(0,x)
    
y = []
for var in x:
    y.append(relu(var))
    
plt.plot(x,y)
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Relu Activaton function: ")
plt.grid()
plt.show()
```
