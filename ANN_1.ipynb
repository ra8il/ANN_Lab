```python
import numpy as np

x = np.array([[0,0],
             [0,1],
             [1,0],
             [1,1]])

y = np.array([0,0,0,1])

weight = [0.1, 0.1]

bias = 0.1

learning_rate = 0.1

epochs = 10
    
def activation(X):
    if(X>=0):
        return 1
    else:
        return 0
    
for epoch in range(epochs):
    print(f"Epochs {epoch + 1}")
    for i in range(len(x)):
        net_input = np.dot(x[i], weight) + bias
        print(net_input)
        pred = activation(net_input)
        print(pred)
        error = y[i] - pred
        weight = weight - learning_rate * error * x[i]
        bias = bias + learning_rate * error
        print(f"Input {x[i]}, Predicted {pred}, Actual {y[i]}")
    
# error = actual_val - observed_val
```
