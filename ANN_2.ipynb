```python
import numpy as np

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])
y = np.array([1,1,1,0])

weights = [0.2, 0.3]
bias = -0.1
lr = 0.1
epochs = 10

def activation(x):
    if x>=0:
        return 1
    else:
        return 0

for epoch in range(epochs):
    print(f"Epoch {epoch + 1}")
    
    for i in range(len(X)):
        net_inp = np.dot(X[i], weights) + bias
        pred = activation(net_inp)
        error = y[i] - pred
        weights + lr * error*X[i]
        bias = bias + lr * error
        print(f"Input: {X[i]}, Target: {y[i]}, Prediction: {pred}")

print("\n Final Weights: ", weights)
print("\n Fianl Bias: ", bias)
```
